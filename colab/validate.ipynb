# Validate dotrepo in Google Colab
Use a standard CPU runtime. Run the cells in order. Google sign-in and Drive authorization are interactive. This checks basic readiness and produces a report; it does not upgrade packages or modify your Drive files.


```python
from pathlib import Path
import subprocess, sys, runpy
repo = Path('/content/dotrepo')
if not repo.exists():
    subprocess.run(['git', 'clone', 'https://github.com/philipjones77/dotrepo.git', str(repo)], check=True)
else:
    print('Using the existing checkout without overwriting local changes.')
subprocess.run(['git', '-C', str(repo), 'log', '-1', '--oneline'], check=True)
```

## Mount Drive and inventory the runtime
Authorize your intended Google account when Colab asks.


```python
runpy.run_path(str(repo / 'colab/setup.py'), run_name='__main__')
```

## Run readiness checks
Dependency conflicts are reported and must be reviewed. CPU-only runtimes are supported.


```python
runpy.run_path(str(repo / 'colab/validate.py'), run_name='__main__')
```
